```python
from thermohl.solver import rte as solver_rte
from thermohl.solver import HeatEquationType
```

```python
# If you define following parameters here they will be ignored:
# - transit
# - ambient_temperature
# - wind_speed
# - solar heating related parameters
# All this is overidden by parameters of solver.reduced_intensity
# or their defaults values.
args = {}
solver = solver_rte(args, heat_equation=HeatEquationType.ONE_TEMPERATURE)

result = solver.reduced_intensity(
    measured_temperature_difference=20,
    measured_intensity=450,
    ambient_temperature=10,
    wind_speed=0.6,
    solar_irradiance=100.0,
    max_conductor_temperature=75,
)

print(f"Reduced intensity: {result} A")
```
